```python
cars=soup.find_all('li',class_='gsc_col-xs-12 gsc_col-sm-6 gsc_col-md-12 gsc_col-lg-12')
```

```python
for i in cars :
  try:
    print(i.find('h3').text)
  except:
    print('not found')
```

```python
for i in cars:
  try:
    print(i.find('div', class_='price').text)
  except:
    print('Price not found')
```

```python
for i in cars :
  try:
    dotlists = i.find_all('div',class_='dotlist')
    for dotlist in dotlists:
        print(dotlist.text)
  except:
    print('not found ')
```
